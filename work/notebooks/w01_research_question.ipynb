
## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.

The starter dataset contains 30,000 content items. 54.21% currently have a downward trend, and 13,152 pages are both trending down and have at least 100 impressions. There are also 9,745 pages with at least 500 impressions, an average position between 1 and 20, and CTR below 0.5%. These numbers suggest there is enough observable opportunity to investigate a ranked review queue, while still requiring volume thresholds to avoid treating small fluctuations as meaningful.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print(f"Rows: {len(df):,}")

declining = (df["trend_direction"] == "down").sum()
declining_with_demand = (
    (df["trend_direction"] == "down") &
    (df["impressions_90d"] >= 100)
).sum()

low_ctr_visible = (
    (df["impressions_90d"] >= 500) &
    (df["avg_position"].between(1, 20)) &
    (df["ctr"] < 0.5)
).sum()

print(f"Declining pages: {declining:,} ({declining / len(df):.2%})")
print(f"Declining + >=100 impressions: {declining_with_demand:,}")
print(f"Visible + position 1-20 + CTR <0.5%: {low_ctr_visible:,}")
```
